```python
from langgraph.graph import  StateGraph, START, END
from typing import TypedDict
from langchain_google_genai import ChatGoogleGenerativeAI
from dotenv import load_dotenv
```

```python
load_dotenv()  # Load environment variables from .env file
model=ChatGoogleGenerativeAI(model="gemini-2.5-flash-preview-05-20")
```

```python
#define state
class LLMState(TypedDict):
    question:str
    answer:str
    
    
```

```python
#define the graph 
graph = StateGraph(LLMState)
```

```python
def llm_qa(state: LLMState) -> LLMState:
    question = state['question']
    prompt=f"Answer the following questions in a concise manner: {question}"
    answer=model.invoke(prompt)
    state['answer']=answer
    return state    
```

```python
graph.add_node('llm_qa',llm_qa)

#add edges 
graph.add_edge(START,'llm_qa')
graph.add_edge('llm_qa',END)

#complieing the graph
result=graph.compile()
```

```python
output=result.invoke({'question':'What is the capital of France?'})
print(output)
```
